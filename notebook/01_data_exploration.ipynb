```python
from sklearn.impute import SimpleImputer

# Create a copy so the original merged dataset stays untouched
model_data = combined_target.copy()

# Impute missing feature values using the median
imputer = SimpleImputer(strategy="median")

model_data[feature_cols] = imputer.fit_transform(
    model_data[feature_cols]
)

print("Missing values after imputation:")
print(model_data[feature_cols].isna().sum())

print("\nFinal shape:", model_data.shape)
```

```python
log_features = ["GDP", "GPC", "CDO", "POP", "UWD"]

for col in log_features:
    model_data[col] = np.log1p(model_data[col])

print(model_data[feature_cols].describe().T)
```

```python
# Indicators we are using
feature_cols = [
    "BHI", "CDA", "CDO", "GDP", "GPC",
    "PMD", "POP", "TCL", "UWD"
]

# Sort chronologically
model_data = model_data.sort_values(["country", "year"])

# Store engineered features
engineered = []

for country, group in model_data.groupby("country"):

    row = {
        "country": country,
        "iso": group["iso"].iloc[0],
        "EPI_score": group["EPI.new"].iloc[0],
        "EPI_rank": group["EPI.rank"].iloc[0]
    }

    for feature in feature_cols:
        values = group[feature]

        # Historical average
        row[f"{feature}_mean"] = values.mean()

        # Latest available value (2023)
        row[f"{feature}_latest"] = values.iloc[-1]

        # Overall trend: linear change per year
        row[f"{feature}_trend"] = np.polyfit(
            group["year"], values, 1
        )[0]

    engineered.append(row)

model_features = pd.DataFrame(engineered)

print("Shape:", model_features.shape)
print(model_features.head())
```

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from xgboost import XGBRegressor

# Train/test split
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

# Proper preprocessing + model pipeline
pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("model", XGBRegressor(
        n_estimators=300,
        max_depth=4,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42
    ))
])

pipeline.fit(X_train, y_train)

y_pred_clean = pipeline.predict(X_test)

print("Pipeline model trained successfully!")
```

```python
# ==========================================
# MODULE 3 — ENVIRONMENTAL TWIN FINDER
# ==========================================

twin_features = [
    "BHI_mean",
    "CDA_mean",
    "CDO_mean",
    "GDP_mean",
    "GPC_mean",
    "PMD_mean",
    "POP_mean",
    "TCL_mean",
    "UWD_mean"
]

twin_data = model_features[
    ["country", "iso"] + twin_features
].copy()

print("Twin Finder dataset shape:", twin_data.shape)
twin_data.head()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_twin = scaler.fit_transform(
    twin_data[twin_features]
)

print("Scaled data shape:", X_twin.shape)
```

```python
X_twin[:5]
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

inertias = []

K_range = range(2, 11)

for k in K_range:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    kmeans.fit(X_twin)
    inertias.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(K_range, inertias, marker="o")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method — Environmental Twin Finder")
plt.xticks(list(K_range))
plt.show()
```

```python
from sklearn.metrics import silhouette_score

silhouette_scores = []

for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    labels = kmeans.fit_predict(X_twin)
    
    score = silhouette_score(X_twin, labels)
    silhouette_scores.append(score)

for k, score in zip(range(2, 11), silhouette_scores):
    print(f"K={k}: Silhouette Score={score:.4f}")
```

```python
from sklearn.cluster import KMeans
import pandas as pd

for k in [2, 4]:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    
    labels = kmeans.fit_predict(X_twin)
    
    print(f"\nK = {k}")
    print(pd.Series(labels).value_counts().sort_index())
```

```python
# Train final Environmental Twin Finder
twin_model = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

twin_data["cluster"] = twin_model.fit_predict(X_twin)

print("Twin Finder trained successfully.")
print("\nCluster sizes:")
print(twin_data["cluster"].value_counts().sort_index())
```

```python
# ==========================================
# MODULE 5 — SUSTAINABILITY SCENARIO ANALYSIS
# ==========================================

def run_scenario(country_name, changes):
    """
    Run a hypothetical sustainability intervention scenario.

    changes:
        Dictionary containing feature changes as percentages.
        Example:
        {"CDA_latest": -10, "UWD_latest": -10}
        
        Negative percentage = reduce the indicator
        Positive percentage = increase the indicator
    """

    # Get country's feature row
    country_row = model_features[
        model_features["country"].str.lower() == country_name.lower()
    ]

    if country_row.empty:
        raise ValueError(f"{country_name} not found.")

    # Extract model features
    scenario = country_row.drop(
        columns=["country", "iso", "EPI_score", "EPI_rank"]
    ).copy()

    # Baseline prediction
    baseline_prediction = pipeline.predict(scenario)[0]

    # Apply hypothetical changes
    for feature, percentage in changes.items():

        if feature not in scenario.columns:
            raise ValueError(
                f"{feature} is not a valid model feature."
            )

        scenario[feature] = (
            scenario[feature] * (1 + percentage / 100)
        )

    # Scenario prediction
    scenario_prediction = pipeline.predict(scenario)[0]

    # Calculate change
    epi_change = scenario_prediction - baseline_prediction

    return {
        "country": country_name,
        "baseline_prediction": baseline_prediction,
        "scenario_prediction": scenario_prediction,
        "epi_change": epi_change,
        "changes": changes
    }
```

```python
scenarios = {
    "Carbon Emission Reduction": {
        "CDA_latest": -10
    },
    "Air Pollution Reduction": {
        "PMD_latest": -10
    },
    "Water Management Improvement": {
        "UWD_latest": -10
    },
    "Economic Growth Scenario": {
        "GDP_latest": 10
    },
    "Per Capita Consumption Increase": {
        "GPC_latest": 10
    }
}

results = []

for scenario_name, changes in scenarios.items():

    result = run_scenario(
        "India",
        changes
    )

    results.append({
        "Scenario": scenario_name,
        "Baseline_EPI": float(result["baseline_prediction"]),
        "Scenario_EPI": float(result["scenario_prediction"]),
        "EPI_Change": float(result["epi_change"])
    })

scenario_results = pd.DataFrame(results)

scenario_results
```

```python
scenario_levels = [5, 10, 20]

interventions = {
    "Carbon Emission Reduction": {
        "CDA_latest": -1
    },
    "Air Pollution Reduction": {
        "PMD_latest": -1
    },
    "Water Management Improvement": {
        "UWD_latest": -1
    }
}

scenario_results = []

for intervention, base_change in interventions.items():
    
    for level in scenario_levels:
        
        changes = {
            feature: change * level
            for feature, change in base_change.items()
        }
        
        result = run_scenario(
            "India",
            changes
        )
        
        scenario_results.append({
            "Intervention": intervention,
            "Improvement": f"{level}%",
            "Baseline_EPI": float(result["baseline_prediction"]),
            "Scenario_EPI": float(result["scenario_prediction"]),
            "EPI_Change": float(result["epi_change"])
        })

scenario_results_df = pd.DataFrame(scenario_results)

scenario_results_df
```
